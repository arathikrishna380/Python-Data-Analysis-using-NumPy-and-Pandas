```python
import numpy as np
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Temperatures Week 1:", temperatures_w1)
```

```python
print("Shape of the array:", temperatures_w1.shape)
print("Data type of the array:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
temperatures_w1_fah = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_w1_fah)

max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum Temperature:", max_temp)
print("Minimum Temperature:", min_temp)
print("Mean Temperature:", mean_temp)
```

```python
first_three_days = temperatures_w1[:3]
print("First three days:", first_three_days)

weekend = temperatures_w1[-2:]
print("Weekend temperatures:", weekend)

middle_three = temperatures_w1[2:5]
print("Middle three days:", middle_three)
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("2D Temperatures Array:\n", temperatures)
```

```python
print("Shape of 2D array:", temperatures.shape)
print("Data type of 2D array:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

week_1 = temperatures[0]
week_2 = temperatures[1]
print("\nWeek 1 temperatures:", week_1)
print("Week 2 temperatures:", week_2)

weekends_both = temperatures[:, -2:]
print("\nWeekends for both weeks (last two days):\n", weekends_both)
```

```python
import pandas as pd

marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'], name='marks')

print("Original Marks Series:\n", marks)
```

```python
print("1st rank student mark (by position):", marks.iloc[0])

print("\nTop 3 ranks using loc:\n", marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\n3rd rank student mark using iloc:", marks.iloc[2])

print("\nRanks with marks greater than 90:\n", marks[marks > 90])
```

```python
marks['Rank1'] = 100
print("Updated Series (Rank 1 as 100):\n", marks)

marks_modified = marks.drop('Rank5')
print("\nSeries after dropping the last rank:\n", marks_modified)

cgpa = marks / 10
print("\nComputed CGPA:\n", cgpa)
```

```python
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print("Transactions DataFrame:\n", transactions)
```

```python
print("--- Head (First 5 rows) ---")
print(transactions.head())

print("\n--- Tail (Last 5 rows) ---")
print(transactions.tail())

print("\n--- Shape of DataFrame ---")
print(transactions.shape)

print("\n--- Column Names ---")
print(transactions.columns)

print("\n--- Data Types ---")
print(transactions.dtypes)
```

```python
print(transactions[['ProductCategory', 'Amount']])
```

```python
print(transactions.iloc[:, -3:])
```

```python
filtered_transactions = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(filtered_transactions)
```

```python
print(transactions['ProductCategory'].value_counts())
```

```python
print(transactions['Region'].unique())
```

```python
mean_amount = transactions.groupby('Region')['Amount'].mean()
print(mean_amount)
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("Updated Amount for Transaction 102:")
print(transactions[transactions['TransactionID'] == 102])
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print("DataFrame with new 'Discount' column:")
print(transactions.head())
```

```python
transactions = transactions[transactions['TransactionID'] != 109]

print("DataFrame after removing Transaction 109:")
print(transactions)
```

```python
transactions = transactions.drop(columns=['Discount'])

print("Final DataFrame after deleting 'Discount' column:")
print(transactions)
```
